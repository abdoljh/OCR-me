```python
import sys, os
sys.path.insert(0, '/content/')

from header_footer_v3 import strip_pdf, Params
from page_export_v2 import export_pages_as_images, extract_footers_pdf

input_pdf = 'Sample.pdf'
input_file = os.path.splitext(os.path.basename(input_pdf))[0]
output_pdf = input_file+"_no_hf.pdf"
raster_pdf = input_file+"_raster.pdf"

output_dir = input_file
output_zip = input_file+".zip"

output_footers_pdf = input_file + "_footers.pdf"

footer_images_dir = input_file + "_footers_images"
footer_images_zip = input_file + "_footers_images.zip"

try:
    # Quiet by default, 300 DPI, full original width, 50 px top/bottom pad
    strip_pdf(input_pdf, output_pdf)
    print(f"Successfully stripped headers and footers from {input_pdf} and saved to {output_pdf}")

    # OCR-ready flattened raster output
    #strip_pdf(input_pdf, raster_pdf, mode="raster")

except Exception as e:
    print(f"An error occurred during PDF processing: {e}")

# Export cleaned pages as images
exported_images = export_pages_as_images(
    in_pdf=output_pdf,
    out_dir=output_dir,
    dpi=400,
    fmt="png",
    zip_path=output_zip
)

print(f"Exported {len(exported_images)} images to '{output_dir}' and zipped them to '{output_zip}'.")

# Export footers
extracted_pages = extract_footers_pdf(input_pdf, output_footers_pdf)

if extracted_pages:
    print(f"Extracted footers from pages {extracted_pages} and saved to '{output_footers_pdf}'.")
else:
    print(f"No footers were extracted from '{input_pdf}'.")

# Extract footer images
# Ensure output directories are created if they don't exist
os.makedirs(footer_images_dir, exist_ok=True)

try:
    extracted_footer_images = extract_footers_pdf(
        src_pdf=input_pdf,
        out_pdf=output_footers_pdf, # This was already created, but needs to be passed
        img_dir=footer_images_dir,
        images_dpi=400,
        images_fmt="png",
        zip_path=footer_images_zip
    )
    if extracted_footer_images:
        print(f"Successfully extracted footer images from pages {extracted_footer_images} to '{footer_images_dir}' and zipped them to '{footer_images_zip}'.")
    else:
        print(f"No footer images were extracted from '{input_pdf}'.")
except Exception as e:
    print(f"An error occurred during footer image extraction: {e}")
```
